```python
from prototype_data import PrototypeData
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import ipywidgets as widgets
```

```python
# METADATA

num_rows = 1000 #how many records to generate
proto = PrototypeData(locale="en_CA", seed=64) #initialize the PrototypeData class with a seed for reproducibility
```

```python
#NUMERICAL DATA PREP

#Income is typically log-normally distributed (right-skewed)
avg_income = 55000
raw_income = np.random.lognormal(mean=10, sigma=0.5, size=num_rows)
scaled_income = raw_income / raw_income.mean() * avg_income

#Height is typically normally distributed (bell curve)
# loc = mean, scale = standard deviation
heights = np.random.normal(loc=170, scale=10, size=num_rows)

#Test Score is sometimes left-skewed, there are few natural examples (e.g. literacy test)
max_score = 100
avg_score = 85
raw_score =  np.random.lognormal(mean=2, sigma=0.3, size=num_rows)
# Flip to make left-skewed
left_skew_scores = max_score - (raw_score / raw_score.max() * max_score)
# Scale to desired average
scaled_scores = left_skew_scores / left_skew_scores.mean() * avg_score

```

```python
mock_df = pd.DataFrame({
    "Name": [proto.fake_name() for _ in range(num_rows)],
    "Address": [proto.fake_address() for _ in range(num_rows)],
    "Email": [proto.fake_email() for _ in range(num_rows)],
    "Income": np.round(scaled_income, 2),
    "Height_cm": np.round(heights, 1),
    "Literacy_Test_Score": np.round(scaled_scores, 1)
})

# Jupyter auto-display
mock_df
```

```python
#VISUALIZE DATA

# Create dropdown widget with numeric columns only
numeric_cols = mock_df.select_dtypes(include=np.number).columns.tolist()
dropdown = widgets.Dropdown(
    options=numeric_cols,
    value=numeric_cols[0],
    description='Select Column:',
)

# Function to plot histogram
def plot_histogram(column):
    plt.figure(figsize=(10,5))
    sns.histplot(mock_df[column], bins=30, kde=True, color="skyblue")
    plt.title(f"Distribution of {column}")
    plt.xlabel(column)
    plt.ylabel("Count")
    plt.show()

# Link dropdown to plotting function
widgets.interact(plot_histogram, column=dropdown)
```
