```python
#IMPORT LIBRARIES
from prototype_data import PrototypeData
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import ipywidgets as widgets
```

```python
# METADATA

num_rows = 1000 #how many records to generate
proto = PrototypeData(locale="en_CA", seed=47) #initialize the PrototypeData class with a seed for reproducibility
```

```python
# GENERATE MOCK DATA

mock_df = pd.DataFrame({
    "Name": [proto.fake_name() for _ in range(num_rows)],
    "Address": [proto.fake_address() for _ in range(num_rows)],
    "Email": [proto.fake_email() for _ in range(num_rows)],
    "Income": np.round(proto.generate_distribution("right-skew", num_rows, mean=10, sigma=0.5, target_mean=55000), 2),
    "Height_cm": np.round(proto.generate_distribution("normal", num_rows, loc=170, scale=10), 1),
    "Easy_Test_Score": np.round(proto.generate_distribution("left-skew", num_rows, mean=10, sigma=0.5, max_value=100, target_mean=85), 1)
})

# Jupyter auto-display
mock_df
```

```python
#VISUALIZE DATA

# Create dropdown widget with numeric columns only
numeric_cols = mock_df.select_dtypes(include=np.number).columns.tolist()
dropdown = widgets.Dropdown(
    options=numeric_cols,
    value=numeric_cols[0],
    description='Select Column:',
)

# Function to plot histogram
def plot_histogram(column):
    plt.figure(figsize=(10,5))
    sns.histplot(mock_df[column], bins=30, kde=True, color="skyblue")
    plt.title(f"Distribution of {column}")
    plt.xlabel(column)
    plt.ylabel("Count")
    plt.show()

# Link dropdown to plotting function
widgets.interact(plot_histogram, column=dropdown)
```

```python
#EXPORT DATA AS CSV

mock_df.to_csv("data_individuals.csv", index=False)
```
